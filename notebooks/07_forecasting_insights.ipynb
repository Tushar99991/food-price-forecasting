```python
# ============================================================
# 07_FORECASTING_INSIGHTS
# FOOD PRICE FORECASTING — INSIGHTS & BUSINESS ANALYSIS
# ============================================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path.cwd().parent

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

FORECASTING_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "forecasting"
)

EVALUATION_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "evaluation"
)

INSIGHTS_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "insights"
)

INSIGHTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. LOAD DATASETS
# ============================================================

print("=" * 70)
print("FOOD PRICE FORECASTING")
print("FORECASTING INSIGHTS")
print("=" * 70)

print("\nLoading project datasets...")


monthly_prices = pd.read_csv(
    PROCESSED_DIR
    / "monthly_crop_prices.csv"
)

# Normalize the monthly dataset schema because different processed
# notebook versions may save the date as "Date", "Month Start",
# or "Month_Start", and the monthly price as "Price" or
# "Median_Price".
normalized_columns = {
    str(column).strip().lower().replace("_", " "): column
    for column in monthly_prices.columns
}

month_column = None

for candidate in [
    "month start",
    "date",
]:
    if candidate in normalized_columns:
        month_column = normalized_columns[candidate]
        break

if month_column is None:
    raise KeyError(
        "Could not find the monthly date column. Expected a column "
        "such as 'Date', 'Month Start', or 'Month_Start'. Available "
        f"columns: {list(monthly_prices.columns)}"
    )

monthly_prices["Month_Start"] = pd.to_datetime(
    monthly_prices[month_column],
    errors="coerce"
)

# Standardize the monthly target price column.
price_column = None

for candidate in [
    "median price",
    "price",
]:
    if candidate in normalized_columns:
        price_column = normalized_columns[candidate]
        break

if price_column is None:
    raise KeyError(
        "Could not find the monthly price column. Expected either "
        "'Median_Price' or 'Price'. Available columns: "
        f"{list(monthly_prices.columns)}"
    )

if "Median_Price" not in monthly_prices.columns:
    monthly_prices["Median_Price"] = pd.to_numeric(
        monthly_prices[price_column],
        errors="coerce"
    )


features = pd.read_csv(
    PROCESSED_DIR
    / "food_price_forecasting_features.csv"
)

features["Month_Start"] = pd.to_datetime(
    features["Month_Start"],
    errors="coerce"
)


crop_performance = pd.read_csv(
    EVALUATION_DIR
    / "crop_level_performance.csv"
)


price_band_performance = pd.read_csv(
    EVALUATION_DIR
    / "price_band_performance.csv"
)


test_predictions = pd.read_csv(
    FORECASTING_DIR
    / "test_predictions.csv"
)

test_predictions["Month_Start"] = pd.to_datetime(
    test_predictions["Month_Start"],
    errors="coerce"
)


print("\nDatasets loaded successfully.")

print(
    f"\nMonthly price rows:       {len(monthly_prices):,}"
)

print(
    f"Feature rows:             {len(features):,}"
)

print(
    f"Crop evaluation rows:     {len(crop_performance):,}"
)

print(
    f"Test prediction rows:     {len(test_predictions):,}"
)


# ============================================================
# 3. BASIC DATA VALIDATION
# ============================================================

required_monthly_columns = [
    "Crop",
    "Year",
    "Month",
    "Median_Price",
    "Mean_Price",
    "Price_Observations",
    "Markets",
    "States"
]


missing_monthly = [
    col
    for col in required_monthly_columns
    if col not in monthly_prices.columns
]


if missing_monthly:

    raise ValueError(
        f"Missing monthly price columns: "
        f"{missing_monthly}"
    )


monthly_prices = (
    monthly_prices
    .dropna(
        subset=[
            "Crop",
            "Month_Start",
            "Median_Price"
        ]
    )
    .copy()
)


monthly_prices = (
    monthly_prices
    .sort_values(
        ["Crop", "Month_Start"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 4. OVERALL PRICE SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("OVERALL PRICE SUMMARY")
print("=" * 70)


overall_summary = pd.DataFrame(
    {
        "Metric": [
            "Number of crops",
            "Number of monthly observations",
            "First month",
            "Last month",
            "Overall median price",
            "Overall mean price",
            "Minimum monthly median",
            "Maximum monthly median"
        ],
        "Value": [
            monthly_prices["Crop"].nunique(),
            len(monthly_prices),
            monthly_prices["Month_Start"].min(),
            monthly_prices["Month_Start"].max(),
            monthly_prices["Median_Price"].median(),
            monthly_prices["Mean_Price"].mean(),
            monthly_prices["Median_Price"].min(),
            monthly_prices["Median_Price"].max()
        ]
    }
)


display(
    overall_summary
)


overall_summary.to_csv(
    INSIGHTS_DIR
    / "overall_summary.csv",
    index=False
)


# ============================================================
# 5. CROP PRICE SUMMARY
# ============================================================

crop_summary = (
    monthly_prices
    .groupby("Crop")
    .agg(
        Months=("Month_Start", "count"),
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean"),
        Minimum_Price=("Median_Price", "min"),
        Maximum_Price=("Median_Price", "max"),
        Price_Std=("Median_Price", "std"),
        Average_Markets=("Markets", "mean"),
        Average_States=("States", "mean")
    )
    .reset_index()
)


crop_summary["Price_Range"] = (
    crop_summary["Maximum_Price"]
    -
    crop_summary["Minimum_Price"]
)


crop_summary["Coefficient_of_Variation"] = (
    crop_summary["Price_Std"]
    /
    crop_summary["Median_Price"]
)


crop_summary = (
    crop_summary
    .sort_values(
        "Median_Price",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("CROP PRICE SUMMARY")
print("=" * 70)


display(
    crop_summary.round(3)
)


crop_summary.to_csv(
    INSIGHTS_DIR
    / "crop_price_summary.csv",
    index=False
)


# ============================================================
# 6. LONG-TERM PRICE GROWTH
# ============================================================

growth_results = []


for crop in sorted(
    monthly_prices["Crop"].unique()
):

    crop_data = (
        monthly_prices[
            monthly_prices["Crop"] == crop
        ]
        .sort_values("Month_Start")
        .copy()
    )


    if len(crop_data) < 2:
        continue


    first_row = crop_data.iloc[0]
    last_row = crop_data.iloc[-1]


    first_price = float(
        first_row["Median_Price"]
    )

    last_price = float(
        last_row["Median_Price"]
    )


    if first_price != 0:

        growth_pct = (
            (
                last_price
                -
                first_price
            )
            /
            first_price
        ) * 100

    else:

        growth_pct = np.nan


    growth_results.append(
        {
            "Crop": crop,
            "Start_Month": first_row["Month_Start"],
            "End_Month": last_row["Month_Start"],
            "Start_Price": first_price,
            "End_Price": last_price,
            "Absolute_Change": (
                last_price
                -
                first_price
            ),
            "Percentage_Growth": growth_pct,
            "Months_Covered": len(crop_data)
        }
    )


growth_summary = (
    pd.DataFrame(growth_results)
    .sort_values(
        "Percentage_Growth",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("LONG-TERM PRICE GROWTH")
print("=" * 70)


display(
    growth_summary.round(2)
)


growth_summary.to_csv(
    INSIGHTS_DIR
    / "crop_long_term_growth.csv",
    index=False
)


# ============================================================
# 7. YEAR-OVER-YEAR PRICE CHANGE
# ============================================================

yearly_crop_prices = (
    monthly_prices
    .groupby(
        ["Crop", "Year"]
    )
    .agg(
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean"),
        Observations=("Price_Observations", "sum")
    )
    .reset_index()
)


yearly_crop_prices[
    "YoY_Change_Percent"
] = (
    yearly_crop_prices
    .groupby("Crop")["Median_Price"]
    .pct_change()
    * 100
)


yearly_crop_prices[
    "YoY_Change_Absolute"
] = (
    yearly_crop_prices
    .groupby("Crop")["Median_Price"]
    .diff()
)


print("\n" + "=" * 70)
print("YEAR-OVER-YEAR PRICE CHANGES")
print("=" * 70)


display(
    yearly_crop_prices
    .tail(20)
    .round(2)
)


yearly_crop_prices.to_csv(
    INSIGHTS_DIR
    / "yearly_crop_price_changes.csv",
    index=False
)


# ============================================================
# 8. MOST RECENT YEAR-OVER-YEAR CHANGES
# ============================================================

latest_year = (
    yearly_crop_prices["Year"].max()
)

previous_year = latest_year - 1


latest_yoy = (
    yearly_crop_prices[
        yearly_crop_prices["Year"].isin(
            [
                previous_year,
                latest_year
            ]
        )
    ]
    .pivot(
        index="Crop",
        columns="Year",
        values="Median_Price"
    )
    .reset_index()
)


if (
    previous_year in latest_yoy.columns
    and
    latest_year in latest_yoy.columns
):

    latest_yoy[
        "YoY_Change_Percent"
    ] = (
        (
            latest_yoy[latest_year]
            -
            latest_yoy[previous_year]
        )
        /
        latest_yoy[previous_year]
        *
        100
    )


print("\n" + "=" * 70)
print(
    f"LATEST YEAR-OVER-YEAR PRICE CHANGE "
    f"({previous_year} → {latest_year})"
)
print("=" * 70)


display(
    latest_yoy.round(2)
)


latest_yoy.to_csv(
    INSIGHTS_DIR
    / "latest_year_over_year_changes.csv",
    index=False
)


# ============================================================
# 9. PRICE VOLATILITY
# ============================================================

volatility_results = []


for crop in sorted(
    monthly_prices["Crop"].unique()
):

    crop_data = (
        monthly_prices[
            monthly_prices["Crop"] == crop
        ]
        .sort_values("Month_Start")
        .copy()
    )


    crop_data["Monthly_Return"] = (
        crop_data["Median_Price"]
        .pct_change()
        * 100
    )


    returns = (
        crop_data["Monthly_Return"]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
    )


    if len(returns) == 0:
        continue


    volatility_results.append(
        {
            "Crop": crop,
            "Monthly_Return_Std": returns.std(),
            "Mean_Monthly_Return": returns.mean(),
            "Absolute_Mean_Monthly_Return": (
                returns.abs().mean()
            ),
            "Largest_Increase": returns.max(),
            "Largest_Decrease": returns.min(),
            "Return_Observations": len(returns)
        }
    )


volatility_summary = (
    pd.DataFrame(volatility_results)
    .sort_values(
        "Monthly_Return_Std",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("PRICE VOLATILITY")
print("=" * 70)


display(
    volatility_summary.round(3)
)


volatility_summary.to_csv(
    INSIGHTS_DIR
    / "crop_price_volatility.csv",
    index=False
)


# ============================================================
# 10. VOLATILITY VISUALIZATION
# ============================================================

plt.figure(
    figsize=(11, 6)
)


volatility_plot = (
    volatility_summary
    .sort_values("Monthly_Return_Std")
)


sns.barplot(
    data=volatility_plot,
    x="Monthly_Return_Std",
    y="Crop"
)


plt.xlabel(
    "Monthly Price Return Volatility (%)"
)

plt.ylabel(
    "Crop"
)

plt.title(
    "Crop Price Volatility"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "crop_price_volatility.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 11. SEASONALITY ANALYSIS
# ============================================================

seasonal_summary = (
    monthly_prices
    .groupby(
        ["Crop", "Month"]
    )
    .agg(
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean"),
        Observations=("Price_Observations", "sum")
    )
    .reset_index()
)


seasonal_pivot = (
    seasonal_summary
    .pivot(
        index="Crop",
        columns="Month",
        values="Median_Price"
    )
)


print("\n" + "=" * 70)
print("SEASONAL PRICE PATTERNS")
print("=" * 70)


display(
    seasonal_pivot.round(2)
)


seasonal_summary.to_csv(
    INSIGHTS_DIR
    / "seasonal_price_summary.csv",
    index=False
)


seasonal_pivot.to_csv(
    INSIGHTS_DIR
    / "seasonal_price_matrix.csv"
)


# ============================================================
# 12. SEASONALITY INDEX
# ============================================================

seasonality_results = []


for crop in seasonal_summary["Crop"].unique():

    crop_seasonal = (
        seasonal_summary[
            seasonal_summary["Crop"] == crop
        ]
        .copy()
    )


    overall_crop_price = (
        crop_seasonal["Median_Price"]
        .mean()
    )


    crop_seasonal[
        "Seasonality_Index"
    ] = (
        crop_seasonal["Median_Price"]
        /
        overall_crop_price
    )


    highest_month = (
        crop_seasonal
        .sort_values(
            "Seasonality_Index",
            ascending=False
        )
        .iloc[0]
    )


    lowest_month = (
        crop_seasonal
        .sort_values(
            "Seasonality_Index"
        )
        .iloc[0]
    )


    seasonality_results.append(
        {
            "Crop": crop,
            "Highest_Price_Month": int(
                highest_month["Month"]
            ),
            "Highest_Seasonality_Index": (
                highest_month[
                    "Seasonality_Index"
                ]
            ),
            "Lowest_Price_Month": int(
                lowest_month["Month"]
            ),
            "Lowest_Seasonality_Index": (
                lowest_month[
                    "Seasonality_Index"
                ]
            ),
            "Seasonality_Spread": (
                highest_month[
                    "Seasonality_Index"
                ]
                -
                lowest_month[
                    "Seasonality_Index"
                ]
            )
        }
    )


seasonality_summary = (
    pd.DataFrame(seasonality_results)
    .sort_values(
        "Seasonality_Spread",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("SEASONALITY INSIGHTS")
print("=" * 70)


display(
    seasonality_summary.round(3)
)


seasonality_summary.to_csv(
    INSIGHTS_DIR
    / "seasonality_insights.csv",
    index=False
)


# ============================================================
# 13. SEASONAL HEATMAP
# ============================================================

plt.figure(
    figsize=(14, 7)
)


sns.heatmap(
    seasonal_pivot,
    annot=True,
    fmt=".0f",
    linewidths=0.5
)


plt.title(
    "Median Monthly Price by Crop"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Crop"
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "seasonal_price_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 14. MARKET COVERAGE INSIGHTS
# ============================================================

market_summary = (
    monthly_prices
    .groupby("Crop")
    .agg(
        Average_Markets=("Markets", "mean"),
        Maximum_Markets=("Markets", "max"),
        Average_States=("States", "mean"),
        Maximum_States=("States", "max")
    )
    .reset_index()
)


market_summary = (
    market_summary
    .sort_values(
        "Average_Markets",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("MARKET COVERAGE")
print("=" * 70)


display(
    market_summary.round(2)
)


market_summary.to_csv(
    INSIGHTS_DIR
    / "market_coverage_summary.csv",
    index=False
)


# ============================================================
# 15. FORECAST PERFORMANCE BY CROP
# ============================================================

forecast_performance = (
    crop_performance.copy()
)


forecast_performance[
    "Forecast_Error_Ratio"
] = (
    forecast_performance["MAE"]
    /
    (
        forecast_performance["RMSE"]
        +
        1e-9
    )
)


forecast_performance = (
    forecast_performance
    .sort_values("RMSE")
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("FORECAST PERFORMANCE BY CROP")
print("=" * 70)


display(
    forecast_performance.round(4)
)


forecast_performance.to_csv(
    INSIGHTS_DIR
    / "forecast_performance_by_crop.csv",
    index=False
)


# ============================================================
# 16. PRICE CHARACTERISTICS VS FORECAST PERFORMANCE
# ============================================================

forecast_analysis = (
    crop_summary[
        [
            "Crop",
            "Median_Price",
            "Price_Std",
            "Coefficient_of_Variation"
        ]
    ]
    .merge(
        crop_performance[
            [
                "Crop",
                "MAE",
                "RMSE",
                "MAPE",
                "SMAPE",
                "R2"
            ]
        ],
        on="Crop",
        how="inner"
    )
)


forecast_analysis.to_csv(
    INSIGHTS_DIR
    / "forecast_and_price_characteristics.csv",
    index=False
)


print("\n" + "=" * 70)
print("PRICE CHARACTERISTICS VS FORECAST PERFORMANCE")
print("=" * 70)


display(
    forecast_analysis.round(4)
)


# ============================================================
# 17. FORECAST ERROR BY PRICE BAND
# ============================================================

test_predictions["Absolute_Error"] = (
    np.abs(
        test_predictions["Actual_Price"]
        -
        test_predictions["Predicted_Price"]
    )
)


test_predictions["Percentage_Error"] = np.where(
    test_predictions["Actual_Price"] != 0,
    (
        test_predictions["Absolute_Error"]
        /
        test_predictions["Actual_Price"]
    ) * 100,
    np.nan
)


test_predictions[
    "Price_Band"
] = pd.cut(
    test_predictions["Actual_Price"],
    bins=[
        -np.inf,
        2000,
        5000,
        7500,
        10000,
        20000,
        50000,
        100000,
        np.inf
    ],
    labels=[
        "< ₹2,000",
        "₹2,000–₹5,000",
        "₹5,000–₹7,500",
        "₹7,500–₹10,000",
        "₹10,000–₹20,000",
        "₹20,000–₹50,000",
        "₹50,000–₹100,000",
        "≥ ₹100,000"
    ]
)


error_by_band = (
    test_predictions
    .groupby(
        "Price_Band",
        observed=True
    )
    .agg(
        Observations=("Actual_Price", "count"),
        Mean_Actual_Price=("Actual_Price", "mean"),
        Mean_Predicted_Price=("Predicted_Price", "mean"),
        MAE=("Absolute_Error", "mean"),
        Mean_Percentage_Error=(
            "Percentage_Error",
            "mean"
        )
    )
    .reset_index()
)


print("\n" + "=" * 70)
print("FORECAST ERROR BY PRICE BAND")
print("=" * 70)


display(
    error_by_band.round(3)
)


error_by_band.to_csv(
    INSIGHTS_DIR
    / "forecast_error_by_price_band.csv",
    index=False
)


# ============================================================
# 18. LARGEST FORECAST ERRORS
# ============================================================

largest_errors = (
    test_predictions
    .sort_values(
        "Absolute_Error",
        ascending=False
    )
    .head(25)
    [
        [
            "Crop",
            "Month_Start",
            "Actual_Price",
            "Predicted_Price",
            "Absolute_Error",
            "Percentage_Error"
        ]
    ]
)


print("\n" + "=" * 70)
print("LARGEST FORECAST ERRORS")
print("=" * 70)


display(
    largest_errors.round(2)
)


largest_errors.to_csv(
    INSIGHTS_DIR
    / "largest_forecast_errors.csv",
    index=False
)


# ============================================================
# 19. FORECAST BIAS BY CROP
# ============================================================

test_predictions["Residual"] = (
    test_predictions["Actual_Price"]
    -
    test_predictions["Predicted_Price"]
)


bias_by_crop = (
    test_predictions
    .groupby("Crop")
    .agg(
        Observations=("Residual", "count"),
        Mean_Residual=("Residual", "mean"),
        Median_Residual=("Residual", "median"),
        Mean_Absolute_Error=(
            "Absolute_Error",
            "mean"
        )
    )
    .reset_index()
)


bias_by_crop[
    "Bias_Direction"
] = np.where(
    bias_by_crop["Mean_Residual"] < 0,
    "Predictions above actuals",
    "Predictions below actuals"
)


bias_by_crop = (
    bias_by_crop
    .sort_values("Mean_Residual")
)


print("\n" + "=" * 70)
print("FORECAST BIAS BY CROP")
print("=" * 70)


display(
    bias_by_crop.round(2)
)


bias_by_crop.to_csv(
    INSIGHTS_DIR
    / "forecast_bias_by_crop.csv",
    index=False
)


# ============================================================
# 20. OVERALL PRICE TREND
# ============================================================

overall_monthly = (
    monthly_prices
    .groupby("Month_Start")
    .agg(
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean")
    )
    .reset_index()
)


plt.figure(
    figsize=(14, 7)
)


plt.plot(
    overall_monthly["Month_Start"],
    overall_monthly["Median_Price"],
    linewidth=2,
    label="Median Price"
)


plt.plot(
    overall_monthly["Month_Start"],
    overall_monthly["Mean_Price"],
    linewidth=2,
    linestyle="--",
    label="Mean Price"
)


plt.xlabel(
    "Month"
)

plt.ylabel(
    "Price (₹/Quintal)"
)

plt.title(
    "Overall Food Commodity Price Trend"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "overall_price_trend.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


overall_monthly.to_csv(
    INSIGHTS_DIR
    / "overall_monthly_price_trend.csv",
    index=False
)


# ============================================================
# 21. CROP PRICE TREND COMPARISON
# ============================================================

plt.figure(
    figsize=(15, 8)
)


for crop in sorted(
    monthly_prices["Crop"].unique()
):

    crop_data = (
        monthly_prices[
            monthly_prices["Crop"] == crop
        ]
        .sort_values("Month_Start")
    )


    plt.plot(
        crop_data["Month_Start"],
        crop_data["Median_Price"],
        linewidth=1.8,
        label=crop
    )


plt.xlabel(
    "Month"
)

plt.ylabel(
    "Median Price (₹/Quintal)"
)

plt.title(
    "Monthly Median Price Trends by Crop"
)


plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "crop_price_trends.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 22. CROP PRICE DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(14, 7)
)


sns.boxplot(
    data=monthly_prices,
    x="Crop",
    y="Median_Price"
)


plt.xlabel(
    "Crop"
)

plt.ylabel(
    "Monthly Median Price (₹/Quintal)"
)

plt.title(
    "Distribution of Monthly Median Prices by Crop"
)


plt.xticks(
    rotation=35,
    ha="right"
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "crop_price_distribution.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 23. KEY INSIGHT TABLE
# ============================================================

highest_median_crop = (
    crop_summary
    .sort_values(
        "Median_Price",
        ascending=False
    )
    .iloc[0]
)


lowest_median_crop = (
    crop_summary
    .sort_values(
        "Median_Price"
    )
    .iloc[0]
)


most_volatile_crop = (
    volatility_summary
    .sort_values(
        "Monthly_Return_Std",
        ascending=False
    )
    .iloc[0]
)


least_volatile_crop = (
    volatility_summary
    .sort_values(
        "Monthly_Return_Std"
    )
    .iloc[0]
)


highest_growth_crop = (
    growth_summary
    .sort_values(
        "Percentage_Growth",
        ascending=False
    )
    .iloc[0]
)


lowest_growth_crop = (
    growth_summary
    .sort_values(
        "Percentage_Growth"
    )
    .iloc[0]
)


best_forecast_crop = (
    crop_performance
    .sort_values(
        "RMSE"
    )
    .iloc[0]
)


highest_forecast_error_crop = (
    crop_performance
    .sort_values(
        "RMSE",
        ascending=False
    )
    .iloc[0]
)


key_insights = pd.DataFrame(
    {
        "Insight": [
            "Highest median-priced crop",
            "Lowest median-priced crop",
            "Highest long-term growth crop",
            "Lowest long-term growth crop",
            "Most volatile crop",
            "Least volatile crop",
            "Lowest forecast RMSE crop",
            "Highest forecast RMSE crop"
        ],
        "Crop": [
            highest_median_crop["Crop"],
            lowest_median_crop["Crop"],
            highest_growth_crop["Crop"],
            lowest_growth_crop["Crop"],
            most_volatile_crop["Crop"],
            least_volatile_crop["Crop"],
            best_forecast_crop["Crop"],
            highest_forecast_error_crop["Crop"]
        ],
        "Value": [
            highest_median_crop["Median_Price"],
            lowest_median_crop["Median_Price"],
            highest_growth_crop["Percentage_Growth"],
            lowest_growth_crop["Percentage_Growth"],
            most_volatile_crop["Monthly_Return_Std"],
            least_volatile_crop["Monthly_Return_Std"],
            best_forecast_crop["RMSE"],
            highest_forecast_error_crop["RMSE"]
        ]
    }
)


print("\n" + "=" * 70)
print("KEY PROJECT INSIGHTS")
print("=" * 70)


display(
    key_insights.round(3)
)


key_insights.to_csv(
    INSIGHTS_DIR
    / "key_project_insights.csv",
    index=False
)


# ============================================================
# 24. FINAL PORTFOLIO METRICS
# ============================================================

portfolio_metrics = pd.DataFrame(
    {
        "Metric": [
            "Crops analyzed",
            "Monthly crop observations",
            "Historical start",
            "Historical end",
            "Best forecasting model",
            "Test observations",
            "Average crop MAE",
            "Average crop RMSE",
            "Average crop MAPE",
            "Average crop SMAPE",
            "Average crop R2"
        ],
        "Value": [
            monthly_prices["Crop"].nunique(),
            len(monthly_prices),
            monthly_prices["Month_Start"].min(),
            monthly_prices["Month_Start"].max(),
            "Extra_Trees",
            len(test_predictions),
            crop_performance["MAE"].mean(),
            crop_performance["RMSE"].mean(),
            crop_performance["MAPE"].mean(),
            crop_performance["SMAPE"].mean(),
            crop_performance["R2"].mean()
        ]
    }
)


portfolio_metrics.to_csv(
    INSIGHTS_DIR
    / "portfolio_metrics.csv",
    index=False
)


# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FORECASTING INSIGHTS COMPLETE")
print("=" * 70)


print(
    f"\nCrops analyzed: "
    f"{monthly_prices['Crop'].nunique()}"
)


print(
    f"Monthly crop observations: "
    f"{len(monthly_prices):,}"
)


print(
    f"Historical period: "
    f"{monthly_prices['Month_Start'].min():%b %Y}"
    f" → "
    f"{monthly_prices['Month_Start'].max():%b %Y}"
)


print(
    f"\nHighest median-priced crop: "
    f"{highest_median_crop['Crop']}"
)


print(
    f"Most volatile crop: "
    f"{most_volatile_crop['Crop']}"
)


print(
    f"Highest long-term growth crop: "
    f"{highest_growth_crop['Crop']}"
)


print(
    f"\nBest forecast RMSE crop: "
    f"{best_forecast_crop['Crop']}"
)


print(
    f"Best forecast RMSE: "
    f"₹{best_forecast_crop['RMSE']:,.2f}"
)


print(
    f"Highest forecast RMSE crop: "
    f"{highest_forecast_error_crop['Crop']}"
)


print(
    f"Highest forecast RMSE: "
    f"₹{highest_forecast_error_crop['RMSE']:,.2f}"
)


print("\nInsight files saved:")

print(
    f"  ✓ {INSIGHTS_DIR / 'overall_summary.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'crop_price_summary.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'crop_long_term_growth.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'yearly_crop_price_changes.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'latest_year_over_year_changes.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'crop_price_volatility.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'seasonality_insights.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'market_coverage_summary.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_performance_by_crop.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_and_price_characteristics.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_error_by_price_band.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'largest_forecast_errors.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_bias_by_crop.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'key_project_insights.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'portfolio_metrics.csv'}"
)


print("\nInsight charts saved in:")

print(
    f"  ✓ {INSIGHTS_DIR}"
)


print("\n" + "=" * 70)
print("PROJECT ANALYTICS PIPELINE COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# 07_FORECASTING_INSIGHTS
# FOOD PRICE FORECASTING — INSIGHTS & BUSINESS ANALYSIS
# ============================================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path.cwd().parent

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

FORECASTING_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "forecasting"
)

EVALUATION_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "evaluation"
)

INSIGHTS_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "insights"
)

INSIGHTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. LOAD DATASETS
# ============================================================

print("=" * 70)
print("FOOD PRICE FORECASTING")
print("FORECASTING INSIGHTS")
print("=" * 70)

print("\nLoading project datasets...")


monthly_prices = pd.read_csv(
    PROCESSED_DIR
    / "monthly_crop_prices.csv"
)

monthly_prices["Month_Start"] = pd.to_datetime(
    monthly_prices["Month Start"],
    errors="coerce"
)


features = pd.read_csv(
    PROCESSED_DIR
    / "food_price_forecasting_features.csv"
)

features["Month_Start"] = pd.to_datetime(
    features["Month_Start"],
    errors="coerce"
)


crop_performance = pd.read_csv(
    EVALUATION_DIR
    / "crop_level_performance.csv"
)


price_band_performance = pd.read_csv(
    EVALUATION_DIR
    / "price_band_performance.csv"
)


test_predictions = pd.read_csv(
    FORECASTING_DIR
    / "test_predictions.csv"
)

test_predictions["Month_Start"] = pd.to_datetime(
    test_predictions["Month_Start"],
    errors="coerce"
)


print("\nDatasets loaded successfully.")

print(
    f"\nMonthly price rows:       {len(monthly_prices):,}"
)

print(
    f"Feature rows:             {len(features):,}"
)

print(
    f"Crop evaluation rows:     {len(crop_performance):,}"
)

print(
    f"Test prediction rows:     {len(test_predictions):,}"
)


# ============================================================
# 3. BASIC DATA VALIDATION
# ============================================================

required_monthly_columns = [
    "Crop",
    "Year",
    "Month",
    "Median_Price",
    "Mean_Price",
    "Price_Observations",
    "Markets",
    "States"
]


missing_monthly = [
    col
    for col in required_monthly_columns
    if col not in monthly_prices.columns
]


if missing_monthly:

    raise ValueError(
        f"Missing monthly price columns: "
        f"{missing_monthly}"
    )


monthly_prices = (
    monthly_prices
    .dropna(
        subset=[
            "Crop",
            "Month_Start",
            "Median_Price"
        ]
    )
    .copy()
)


monthly_prices = (
    monthly_prices
    .sort_values(
        ["Crop", "Month_Start"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 4. OVERALL PRICE SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("OVERALL PRICE SUMMARY")
print("=" * 70)


overall_summary = pd.DataFrame(
    {
        "Metric": [
            "Number of crops",
            "Number of monthly observations",
            "First month",
            "Last month",
            "Overall median price",
            "Overall mean price",
            "Minimum monthly median",
            "Maximum monthly median"
        ],
        "Value": [
            monthly_prices["Crop"].nunique(),
            len(monthly_prices),
            monthly_prices["Month_Start"].min(),
            monthly_prices["Month_Start"].max(),
            monthly_prices["Median_Price"].median(),
            monthly_prices["Mean_Price"].mean(),
            monthly_prices["Median_Price"].min(),
            monthly_prices["Median_Price"].max()
        ]
    }
)


display(
    overall_summary
)


overall_summary.to_csv(
    INSIGHTS_DIR
    / "overall_summary.csv",
    index=False
)


# ============================================================
# 5. CROP PRICE SUMMARY
# ============================================================

crop_summary = (
    monthly_prices
    .groupby("Crop")
    .agg(
        Months=("Month_Start", "count"),
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean"),
        Minimum_Price=("Median_Price", "min"),
        Maximum_Price=("Median_Price", "max"),
        Price_Std=("Median_Price", "std"),
        Average_Markets=("Markets", "mean"),
        Average_States=("States", "mean")
    )
    .reset_index()
)


crop_summary["Price_Range"] = (
    crop_summary["Maximum_Price"]
    -
    crop_summary["Minimum_Price"]
)


crop_summary["Coefficient_of_Variation"] = (
    crop_summary["Price_Std"]
    /
    crop_summary["Median_Price"]
)


crop_summary = (
    crop_summary
    .sort_values(
        "Median_Price",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("CROP PRICE SUMMARY")
print("=" * 70)


display(
    crop_summary.round(3)
)


crop_summary.to_csv(
    INSIGHTS_DIR
    / "crop_price_summary.csv",
    index=False
)


# ============================================================
# 6. LONG-TERM PRICE GROWTH
# ============================================================

growth_results = []


for crop in sorted(
    monthly_prices["Crop"].unique()
):

    crop_data = (
        monthly_prices[
            monthly_prices["Crop"] == crop
        ]
        .sort_values("Month_Start")
        .copy()
    )


    if len(crop_data) < 2:
        continue


    first_row = crop_data.iloc[0]
    last_row = crop_data.iloc[-1]


    first_price = float(
        first_row["Median_Price"]
    )

    last_price = float(
        last_row["Median_Price"]
    )


    if first_price != 0:

        growth_pct = (
            (
                last_price
                -
                first_price
            )
            /
            first_price
        ) * 100

    else:

        growth_pct = np.nan


    growth_results.append(
        {
            "Crop": crop,
            "Start_Month": first_row["Month_Start"],
            "End_Month": last_row["Month_Start"],
            "Start_Price": first_price,
            "End_Price": last_price,
            "Absolute_Change": (
                last_price
                -
                first_price
            ),
            "Percentage_Growth": growth_pct,
            "Months_Covered": len(crop_data)
        }
    )


growth_summary = (
    pd.DataFrame(growth_results)
    .sort_values(
        "Percentage_Growth",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("LONG-TERM PRICE GROWTH")
print("=" * 70)


display(
    growth_summary.round(2)
)


growth_summary.to_csv(
    INSIGHTS_DIR
    / "crop_long_term_growth.csv",
    index=False
)


# ============================================================
# 7. YEAR-OVER-YEAR PRICE CHANGE
# ============================================================

yearly_crop_prices = (
    monthly_prices
    .groupby(
        ["Crop", "Year"]
    )
    .agg(
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean"),
        Observations=("Price_Observations", "sum")
    )
    .reset_index()
)


yearly_crop_prices[
    "YoY_Change_Percent"
] = (
    yearly_crop_prices
    .groupby("Crop")["Median_Price"]
    .pct_change()
    * 100
)


yearly_crop_prices[
    "YoY_Change_Absolute"
] = (
    yearly_crop_prices
    .groupby("Crop")["Median_Price"]
    .diff()
)


print("\n" + "=" * 70)
print("YEAR-OVER-YEAR PRICE CHANGES")
print("=" * 70)


display(
    yearly_crop_prices
    .tail(20)
    .round(2)
)


yearly_crop_prices.to_csv(
    INSIGHTS_DIR
    / "yearly_crop_price_changes.csv",
    index=False
)


# ============================================================
# 8. MOST RECENT YEAR-OVER-YEAR CHANGES
# ============================================================

latest_year = (
    yearly_crop_prices["Year"].max()
)

previous_year = latest_year - 1


latest_yoy = (
    yearly_crop_prices[
        yearly_crop_prices["Year"].isin(
            [
                previous_year,
                latest_year
            ]
        )
    ]
    .pivot(
        index="Crop",
        columns="Year",
        values="Median_Price"
    )
    .reset_index()
)


if (
    previous_year in latest_yoy.columns
    and
    latest_year in latest_yoy.columns
):

    latest_yoy[
        "YoY_Change_Percent"
    ] = (
        (
            latest_yoy[latest_year]
            -
            latest_yoy[previous_year]
        )
        /
        latest_yoy[previous_year]
        *
        100
    )


print("\n" + "=" * 70)
print(
    f"LATEST YEAR-OVER-YEAR PRICE CHANGE "
    f"({previous_year} → {latest_year})"
)
print("=" * 70)


display(
    latest_yoy.round(2)
)


latest_yoy.to_csv(
    INSIGHTS_DIR
    / "latest_year_over_year_changes.csv",
    index=False
)


# ============================================================
# 9. PRICE VOLATILITY
# ============================================================

volatility_results = []


for crop in sorted(
    monthly_prices["Crop"].unique()
):

    crop_data = (
        monthly_prices[
            monthly_prices["Crop"] == crop
        ]
        .sort_values("Month_Start")
        .copy()
    )


    crop_data["Monthly_Return"] = (
        crop_data["Median_Price"]
        .pct_change()
        * 100
    )


    returns = (
        crop_data["Monthly_Return"]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
    )


    if len(returns) == 0:
        continue


    volatility_results.append(
        {
            "Crop": crop,
            "Monthly_Return_Std": returns.std(),
            "Mean_Monthly_Return": returns.mean(),
            "Absolute_Mean_Monthly_Return": (
                returns.abs().mean()
            ),
            "Largest_Increase": returns.max(),
            "Largest_Decrease": returns.min(),
            "Return_Observations": len(returns)
        }
    )


volatility_summary = (
    pd.DataFrame(volatility_results)
    .sort_values(
        "Monthly_Return_Std",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("PRICE VOLATILITY")
print("=" * 70)


display(
    volatility_summary.round(3)
)


volatility_summary.to_csv(
    INSIGHTS_DIR
    / "crop_price_volatility.csv",
    index=False
)


# ============================================================
# 10. VOLATILITY VISUALIZATION
# ============================================================

plt.figure(
    figsize=(11, 6)
)


volatility_plot = (
    volatility_summary
    .sort_values("Monthly_Return_Std")
)


sns.barplot(
    data=volatility_plot,
    x="Monthly_Return_Std",
    y="Crop"
)


plt.xlabel(
    "Monthly Price Return Volatility (%)"
)

plt.ylabel(
    "Crop"
)

plt.title(
    "Crop Price Volatility"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "crop_price_volatility.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 11. SEASONALITY ANALYSIS
# ============================================================

seasonal_summary = (
    monthly_prices
    .groupby(
        ["Crop", "Month"]
    )
    .agg(
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean"),
        Observations=("Price_Observations", "sum")
    )
    .reset_index()
)


seasonal_pivot = (
    seasonal_summary
    .pivot(
        index="Crop",
        columns="Month",
        values="Median_Price"
    )
)


print("\n" + "=" * 70)
print("SEASONAL PRICE PATTERNS")
print("=" * 70)


display(
    seasonal_pivot.round(2)
)


seasonal_summary.to_csv(
    INSIGHTS_DIR
    / "seasonal_price_summary.csv",
    index=False
)


seasonal_pivot.to_csv(
    INSIGHTS_DIR
    / "seasonal_price_matrix.csv"
)


# ============================================================
# 12. SEASONALITY INDEX
# ============================================================

seasonality_results = []


for crop in seasonal_summary["Crop"].unique():

    crop_seasonal = (
        seasonal_summary[
            seasonal_summary["Crop"] == crop
        ]
        .copy()
    )


    overall_crop_price = (
        crop_seasonal["Median_Price"]
        .mean()
    )


    crop_seasonal[
        "Seasonality_Index"
    ] = (
        crop_seasonal["Median_Price"]
        /
        overall_crop_price
    )


    highest_month = (
        crop_seasonal
        .sort_values(
            "Seasonality_Index",
            ascending=False
        )
        .iloc[0]
    )


    lowest_month = (
        crop_seasonal
        .sort_values(
            "Seasonality_Index"
        )
        .iloc[0]
    )


    seasonality_results.append(
        {
            "Crop": crop,
            "Highest_Price_Month": int(
                highest_month["Month"]
            ),
            "Highest_Seasonality_Index": (
                highest_month[
                    "Seasonality_Index"
                ]
            ),
            "Lowest_Price_Month": int(
                lowest_month["Month"]
            ),
            "Lowest_Seasonality_Index": (
                lowest_month[
                    "Seasonality_Index"
                ]
            ),
            "Seasonality_Spread": (
                highest_month[
                    "Seasonality_Index"
                ]
                -
                lowest_month[
                    "Seasonality_Index"
                ]
            )
        }
    )


seasonality_summary = (
    pd.DataFrame(seasonality_results)
    .sort_values(
        "Seasonality_Spread",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("SEASONALITY INSIGHTS")
print("=" * 70)


display(
    seasonality_summary.round(3)
)


seasonality_summary.to_csv(
    INSIGHTS_DIR
    / "seasonality_insights.csv",
    index=False
)


# ============================================================
# 13. SEASONAL HEATMAP
# ============================================================

plt.figure(
    figsize=(14, 7)
)


sns.heatmap(
    seasonal_pivot,
    annot=True,
    fmt=".0f",
    linewidths=0.5
)


plt.title(
    "Median Monthly Price by Crop"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Crop"
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "seasonal_price_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 14. MARKET COVERAGE INSIGHTS
# ============================================================

market_summary = (
    monthly_prices
    .groupby("Crop")
    .agg(
        Average_Markets=("Markets", "mean"),
        Maximum_Markets=("Markets", "max"),
        Average_States=("States", "mean"),
        Maximum_States=("States", "max")
    )
    .reset_index()
)


market_summary = (
    market_summary
    .sort_values(
        "Average_Markets",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("MARKET COVERAGE")
print("=" * 70)


display(
    market_summary.round(2)
)


market_summary.to_csv(
    INSIGHTS_DIR
    / "market_coverage_summary.csv",
    index=False
)


# ============================================================
# 15. FORECAST PERFORMANCE BY CROP
# ============================================================

forecast_performance = (
    crop_performance.copy()
)


forecast_performance[
    "Forecast_Error_Ratio"
] = (
    forecast_performance["MAE"]
    /
    (
        forecast_performance["RMSE"]
        +
        1e-9
    )
)


forecast_performance = (
    forecast_performance
    .sort_values("RMSE")
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("FORECAST PERFORMANCE BY CROP")
print("=" * 70)


display(
    forecast_performance.round(4)
)


forecast_performance.to_csv(
    INSIGHTS_DIR
    / "forecast_performance_by_crop.csv",
    index=False
)


# ============================================================
# 16. PRICE CHARACTERISTICS VS FORECAST PERFORMANCE
# ============================================================

forecast_analysis = (
    crop_summary[
        [
            "Crop",
            "Median_Price",
            "Price_Std",
            "Coefficient_of_Variation"
        ]
    ]
    .merge(
        crop_performance[
            [
                "Crop",
                "MAE",
                "RMSE",
                "MAPE",
                "SMAPE",
                "R2"
            ]
        ],
        on="Crop",
        how="inner"
    )
)


forecast_analysis.to_csv(
    INSIGHTS_DIR
    / "forecast_and_price_characteristics.csv",
    index=False
)


print("\n" + "=" * 70)
print("PRICE CHARACTERISTICS VS FORECAST PERFORMANCE")
print("=" * 70)


display(
    forecast_analysis.round(4)
)


# ============================================================
# 17. FORECAST ERROR BY PRICE BAND
# ============================================================

test_predictions["Absolute_Error"] = (
    np.abs(
        test_predictions["Actual_Price"]
        -
        test_predictions["Predicted_Price"]
    )
)


test_predictions["Percentage_Error"] = np.where(
    test_predictions["Actual_Price"] != 0,
    (
        test_predictions["Absolute_Error"]
        /
        test_predictions["Actual_Price"]
    ) * 100,
    np.nan
)


test_predictions[
    "Price_Band"
] = pd.cut(
    test_predictions["Actual_Price"],
    bins=[
        -np.inf,
        2000,
        5000,
        7500,
        10000,
        20000,
        50000,
        100000,
        np.inf
    ],
    labels=[
        "< ₹2,000",
        "₹2,000–₹5,000",
        "₹5,000–₹7,500",
        "₹7,500–₹10,000",
        "₹10,000–₹20,000",
        "₹20,000–₹50,000",
        "₹50,000–₹100,000",
        "≥ ₹100,000"
    ]
)


error_by_band = (
    test_predictions
    .groupby(
        "Price_Band",
        observed=True
    )
    .agg(
        Observations=("Actual_Price", "count"),
        Mean_Actual_Price=("Actual_Price", "mean"),
        Mean_Predicted_Price=("Predicted_Price", "mean"),
        MAE=("Absolute_Error", "mean"),
        Mean_Percentage_Error=(
            "Percentage_Error",
            "mean"
        )
    )
    .reset_index()
)


print("\n" + "=" * 70)
print("FORECAST ERROR BY PRICE BAND")
print("=" * 70)


display(
    error_by_band.round(3)
)


error_by_band.to_csv(
    INSIGHTS_DIR
    / "forecast_error_by_price_band.csv",
    index=False
)


# ============================================================
# 18. LARGEST FORECAST ERRORS
# ============================================================

largest_errors = (
    test_predictions
    .sort_values(
        "Absolute_Error",
        ascending=False
    )
    .head(25)
    [
        [
            "Crop",
            "Month_Start",
            "Actual_Price",
            "Predicted_Price",
            "Absolute_Error",
            "Percentage_Error"
        ]
    ]
)


print("\n" + "=" * 70)
print("LARGEST FORECAST ERRORS")
print("=" * 70)


display(
    largest_errors.round(2)
)


largest_errors.to_csv(
    INSIGHTS_DIR
    / "largest_forecast_errors.csv",
    index=False
)


# ============================================================
# 19. FORECAST BIAS BY CROP
# ============================================================

test_predictions["Residual"] = (
    test_predictions["Actual_Price"]
    -
    test_predictions["Predicted_Price"]
)


bias_by_crop = (
    test_predictions
    .groupby("Crop")
    .agg(
        Observations=("Residual", "count"),
        Mean_Residual=("Residual", "mean"),
        Median_Residual=("Residual", "median"),
        Mean_Absolute_Error=(
            "Absolute_Error",
            "mean"
        )
    )
    .reset_index()
)


bias_by_crop[
    "Bias_Direction"
] = np.where(
    bias_by_crop["Mean_Residual"] < 0,
    "Predictions above actuals",
    "Predictions below actuals"
)


bias_by_crop = (
    bias_by_crop
    .sort_values("Mean_Residual")
)


print("\n" + "=" * 70)
print("FORECAST BIAS BY CROP")
print("=" * 70)


display(
    bias_by_crop.round(2)
)


bias_by_crop.to_csv(
    INSIGHTS_DIR
    / "forecast_bias_by_crop.csv",
    index=False
)


# ============================================================
# 20. OVERALL PRICE TREND
# ============================================================

overall_monthly = (
    monthly_prices
    .groupby("Month_Start")
    .agg(
        Median_Price=("Median_Price", "median"),
        Mean_Price=("Mean_Price", "mean")
    )
    .reset_index()
)


plt.figure(
    figsize=(14, 7)
)


plt.plot(
    overall_monthly["Month_Start"],
    overall_monthly["Median_Price"],
    linewidth=2,
    label="Median Price"
)


plt.plot(
    overall_monthly["Month_Start"],
    overall_monthly["Mean_Price"],
    linewidth=2,
    linestyle="--",
    label="Mean Price"
)


plt.xlabel(
    "Month"
)

plt.ylabel(
    "Price (₹/Quintal)"
)

plt.title(
    "Overall Food Commodity Price Trend"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "overall_price_trend.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


overall_monthly.to_csv(
    INSIGHTS_DIR
    / "overall_monthly_price_trend.csv",
    index=False
)


# ============================================================
# 21. CROP PRICE TREND COMPARISON
# ============================================================

plt.figure(
    figsize=(15, 8)
)


for crop in sorted(
    monthly_prices["Crop"].unique()
):

    crop_data = (
        monthly_prices[
            monthly_prices["Crop"] == crop
        ]
        .sort_values("Month_Start")
    )


    plt.plot(
        crop_data["Month_Start"],
        crop_data["Median_Price"],
        linewidth=1.8,
        label=crop
    )


plt.xlabel(
    "Month"
)

plt.ylabel(
    "Median Price (₹/Quintal)"
)

plt.title(
    "Monthly Median Price Trends by Crop"
)


plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "crop_price_trends.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 22. CROP PRICE DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(14, 7)
)


sns.boxplot(
    data=monthly_prices,
    x="Crop",
    y="Median_Price"
)


plt.xlabel(
    "Crop"
)

plt.ylabel(
    "Monthly Median Price (₹/Quintal)"
)

plt.title(
    "Distribution of Monthly Median Prices by Crop"
)


plt.xticks(
    rotation=35,
    ha="right"
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    INSIGHTS_DIR
    / "crop_price_distribution.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 23. KEY INSIGHT TABLE
# ============================================================

highest_median_crop = (
    crop_summary
    .sort_values(
        "Median_Price",
        ascending=False
    )
    .iloc[0]
)


lowest_median_crop = (
    crop_summary
    .sort_values(
        "Median_Price"
    )
    .iloc[0]
)


most_volatile_crop = (
    volatility_summary
    .sort_values(
        "Monthly_Return_Std",
        ascending=False
    )
    .iloc[0]
)


least_volatile_crop = (
    volatility_summary
    .sort_values(
        "Monthly_Return_Std"
    )
    .iloc[0]
)


highest_growth_crop = (
    growth_summary
    .sort_values(
        "Percentage_Growth",
        ascending=False
    )
    .iloc[0]
)


lowest_growth_crop = (
    growth_summary
    .sort_values(
        "Percentage_Growth"
    )
    .iloc[0]
)


best_forecast_crop = (
    crop_performance
    .sort_values(
        "RMSE"
    )
    .iloc[0]
)


highest_forecast_error_crop = (
    crop_performance
    .sort_values(
        "RMSE",
        ascending=False
    )
    .iloc[0]
)


key_insights = pd.DataFrame(
    {
        "Insight": [
            "Highest median-priced crop",
            "Lowest median-priced crop",
            "Highest long-term growth crop",
            "Lowest long-term growth crop",
            "Most volatile crop",
            "Least volatile crop",
            "Lowest forecast RMSE crop",
            "Highest forecast RMSE crop"
        ],
        "Crop": [
            highest_median_crop["Crop"],
            lowest_median_crop["Crop"],
            highest_growth_crop["Crop"],
            lowest_growth_crop["Crop"],
            most_volatile_crop["Crop"],
            least_volatile_crop["Crop"],
            best_forecast_crop["Crop"],
            highest_forecast_error_crop["Crop"]
        ],
        "Value": [
            highest_median_crop["Median_Price"],
            lowest_median_crop["Median_Price"],
            highest_growth_crop["Percentage_Growth"],
            lowest_growth_crop["Percentage_Growth"],
            most_volatile_crop["Monthly_Return_Std"],
            least_volatile_crop["Monthly_Return_Std"],
            best_forecast_crop["RMSE"],
            highest_forecast_error_crop["RMSE"]
        ]
    }
)


print("\n" + "=" * 70)
print("KEY PROJECT INSIGHTS")
print("=" * 70)


display(
    key_insights.round(3)
)


key_insights.to_csv(
    INSIGHTS_DIR
    / "key_project_insights.csv",
    index=False
)


# ============================================================
# 24. FINAL PORTFOLIO METRICS
# ============================================================

portfolio_metrics = pd.DataFrame(
    {
        "Metric": [
            "Crops analyzed",
            "Monthly crop observations",
            "Historical start",
            "Historical end",
            "Best forecasting model",
            "Test observations",
            "Average crop MAE",
            "Average crop RMSE",
            "Average crop MAPE",
            "Average crop SMAPE",
            "Average crop R2"
        ],
        "Value": [
            monthly_prices["Crop"].nunique(),
            len(monthly_prices),
            monthly_prices["Month_Start"].min(),
            monthly_prices["Month_Start"].max(),
            "Extra_Trees",
            len(test_predictions),
            crop_performance["MAE"].mean(),
            crop_performance["RMSE"].mean(),
            crop_performance["MAPE"].mean(),
            crop_performance["SMAPE"].mean(),
            crop_performance["R2"].mean()
        ]
    }
)


portfolio_metrics.to_csv(
    INSIGHTS_DIR
    / "portfolio_metrics.csv",
    index=False
)


# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FORECASTING INSIGHTS COMPLETE")
print("=" * 70)


print(
    f"\nCrops analyzed: "
    f"{monthly_prices['Crop'].nunique()}"
)


print(
    f"Monthly crop observations: "
    f"{len(monthly_prices):,}"
)


print(
    f"Historical period: "
    f"{monthly_prices['Month_Start'].min():%b %Y}"
    f" → "
    f"{monthly_prices['Month_Start'].max():%b %Y}"
)


print(
    f"\nHighest median-priced crop: "
    f"{highest_median_crop['Crop']}"
)


print(
    f"Most volatile crop: "
    f"{most_volatile_crop['Crop']}"
)


print(
    f"Highest long-term growth crop: "
    f"{highest_growth_crop['Crop']}"
)


print(
    f"\nBest forecast RMSE crop: "
    f"{best_forecast_crop['Crop']}"
)


print(
    f"Best forecast RMSE: "
    f"₹{best_forecast_crop['RMSE']:,.2f}"
)


print(
    f"Highest forecast RMSE crop: "
    f"{highest_forecast_error_crop['Crop']}"
)


print(
    f"Highest forecast RMSE: "
    f"₹{highest_forecast_error_crop['RMSE']:,.2f}"
)


print("\nInsight files saved:")

print(
    f"  ✓ {INSIGHTS_DIR / 'overall_summary.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'crop_price_summary.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'crop_long_term_growth.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'yearly_crop_price_changes.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'latest_year_over_year_changes.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'crop_price_volatility.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'seasonality_insights.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'market_coverage_summary.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_performance_by_crop.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_and_price_characteristics.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_error_by_price_band.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'largest_forecast_errors.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'forecast_bias_by_crop.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'key_project_insights.csv'}"
)

print(
    f"  ✓ {INSIGHTS_DIR / 'portfolio_metrics.csv'}"
)


print("\nInsight charts saved in:")

print(
    f"  ✓ {INSIGHTS_DIR}"
)


print("\n" + "=" * 70)
print("PROJECT ANALYTICS PIPELINE COMPLETE")
print("=" * 70)
```
